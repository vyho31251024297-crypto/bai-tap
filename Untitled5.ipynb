```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
import ipywidgets as widgets
from IPython.display import display, clear_output



dulieutieuduong = pd.read_csv('diabetes.csv')
cacchiso = dulieutieuduong.drop('Outcome', axis=1)
ketquabenh = dulieutieuduong['Outcome']
chisohoc, chisokiemtra, ketquahoc, ketquakiemtra = train_test_split(cacchiso, ketquabenh, test_size=0.2, random_state=42)
bochuanhoa = StandardScaler()
chisohocdascale = bochuanhoa.fit_transform(chisohoc)
mohinhchuascale = Perceptron(max_iter=1000, eta0=0.1, random_state=42)
mohinhchuascale.fit(chisohoc, ketquahoc)
mohinhdascale = Perceptron(max_iter=1000, eta0=0.1, random_state=42)
mohinhdascale.fit(chisohocdascale, ketquahoc)
thanhmangthai = widgets.IntSlider(value=1, min=0, max=20, description='Số lần mang thai:')
thanhduonghuyet = widgets.IntSlider(value=100, min=0, max=250, description='Chỉ số Glucose:')
thanhhuyetap = widgets.IntSlider(value=70, min=0, max=150, description='Huyết áp tâm trương:')
thanhdodayda = widgets.IntSlider(value=20, min=0, max=100, description='Độ dày nếp gấp da:')
thanhinsulin = widgets.IntSlider(value=80, min=0, max=900, step=5, description='Chỉ số Insulin:')
thanhbmi = widgets.FloatSlider(value=25.0, min=0.0, max=70.0, step=0.1, description='Chỉ số BMI:')
thanhphahe = widgets.FloatSlider(value=0.5, min=0.0, max=3.0, step=0.01, description='Phả hệ tiểu đường:')
thanhtuoi = widgets.IntSlider(value=30, min=1, max=120, description='Tuổi của bạn:')
luachonmohinh = widgets.Dropdown(
    options=[('Mô hình Đã chuẩn hóa (Nên dùng)', 'dascale'), ('Mô hình Chưa chuẩn hóa', 'chuascale')],
    value='dascale',
    description='Chọn mô hình:'
)
nutbamtinh = widgets.Button(description='Tính toán nguy cơ', button_style='success')
vunghienthiketqua = widgets.Output()
def xulydudoan(b):
    with vunghienthiketqua:
        clear_output()
        thongtinnguoidung = pd.DataFrame([{'Pregnancies': thanhmangthai.value,'Glucose': thanhduonghuyet.value,'BloodPressure': thanhhuyetap.value,'SkinThickness': thanhdodayda.value,'Insulin': thanhinsulin.value,'BMI': thanhbmi.value,'DiabetesPedigreeFunction': thanhphahe.value,'Age': thanhtuoi.value}])
        if luachonmohinh.value == 'chuascale':
            dudoan = mohinhchuascale.predict(thongtinnguoidung)[0]
            dochinhxac = accuracy_score(ketquakiemtra, mohinhchuascale.predict(chisokiemtra))
        else:
            thongtindascale = bochuanhoa.transform(thongtinnguoidung)
            dudoan = mohinhdascale.predict(thongtindascale)[0]
            dochinhxac = accuracy_score(ketquakiemtra, mohinhdascale.predict(bochuanhoa.transform(chisokiemtra)))
        print(f"-> Độ chính xác của mô hình hiện tại: {dochinhxac * 100:.2f}%")
        print("-" * 60)
        if dudoan == 1:
            print("bạn có nguy cơ đã  mắc bệnh tiểu đường.")
        else:
            print("yên tâm an toàn rồi  .")
nutbamtinh.on_click(xulydudoan)
giaodienhoanchinh = widgets.VBox([
    thanhmangthai, thanhduonghuyet, thanhhuyetap, thanhdodayda,
    thanhinsulin, thanhbmi, thanhphahe, thanhtuoi,
    luachonmohinh, nutbamtinh, vunghienthiketqua
])
display(giaodienhoanchinh)
```
