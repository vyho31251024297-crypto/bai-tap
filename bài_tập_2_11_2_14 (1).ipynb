```python
#2.11
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

distance = ctrl.Antecedent(np.arange(0, 51, 1), 'distance')
traffic = ctrl.Antecedent(np.arange(0, 101, 1), 'traffic')
demand = ctrl.Antecedent(np.arange(0, 101, 1), 'demand')
weather = ctrl.Antecedent(np.arange(0, 11, 1), 'weather')


price = ctrl.Consequent(np.arange(0, 201, 1), 'price')

distance['ngan'] = fuzz.trimf(distance.universe, [0, 0, 3])
distance['trung_binh'] = fuzz.trimf(distance.universe, [2, 5, 8])
distance['dai'] = fuzz.trimf(distance.universe, [6, 13, 20])
distance['rat_xa'] = fuzz.trimf(distance.universe, [15, 50, 50])

traffic['thap'] = fuzz.trimf(traffic.universe, [0, 0, 30])
traffic['trung_binh'] = fuzz.trimf(traffic.universe, [20, 46, 70])
traffic['cao'] = fuzz.trimf(traffic.universe, [60, 100, 100])

demand['thap'] = fuzz.trimf(demand.universe, [0, 0, 30])
demand['trung_binh'] = fuzz.trimf(demand.universe, [20, 46, 70])
demand['cao'] = fuzz.trimf(demand.universe, [60, 100, 100])

weather['tot'] = fuzz.trimf(weather.universe, [0, 0, 3])
weather['vua'] = fuzz.trimf(weather.universe, [2, 5, 7])
weather['xau'] = fuzz.trimf(weather.universe, [6, 10, 10])

price['thap'] = fuzz.trimf(price.universe, [0, 30, 70])
price['trung_binh'] = fuzz.trimf(price.universe, [60, 100, 140])
price['cao'] = fuzz.trimf(price.universe, [130, 160, 180])
price['rat_cao'] = fuzz.trimf(price.universe, [170, 200, 200])

rule1 = ctrl.Rule(distance['ngan'] & traffic['thap'] & demand['thap'], price['thap'])
rule2 = ctrl.Rule(distance['ngan'] & traffic['trung_binh'] & demand['cao'], price['trung_binh'])
rule3 = ctrl.Rule(distance['trung_binh'] & traffic['cao'] & demand['cao'], price['cao'])
rule4 = ctrl.Rule(distance['dai'] & traffic['trung_binh'] & weather['tot'], price['trung_binh'])
rule5 = ctrl.Rule(distance['dai'] & traffic['cao'] & weather['xau'], price['rat_cao'])
rule6 = ctrl.Rule(distance['rat_xa'] & traffic['cao'] & demand['cao'], price['rat_cao'])
rule7 = ctrl.Rule(distance['trung_binh'] & traffic['thap'] & demand['thap'], price['trung_binh'])
rule8 = ctrl.Rule(distance['ngan'] & traffic['cao'] & weather['xau'], price['cao'])
rule9 = ctrl.Rule(distance['rat_xa'] & weather['xau'], price['rat_cao'])
rule10 = ctrl.Rule(distance['trung_binh'] & traffic['trung_binh'] & weather['vua'], price['trung_binh'])

grab_ctrl = ctrl.ControlSystem([
    rule1, rule2, rule3, rule4, rule5,
    rule6, rule7, rule8, rule9, rule10 ])

grab_sim = ctrl.ControlSystemSimulation(grab_ctrl)

grab_sim.input['distance'] = 20
grab_sim.input['traffic'] = 90
grab_sim.input['demand'] = 90
grab_sim.input['weather'] = 7

grab_sim.compute()
print("Gia de xuat: ", round(grab_sim.output['price'],2))
```

```python
#2.12
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
#input
rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'rating')
sales = ctrl.Antecedent(np.arange(0, 1001, 1), 'sales')
margin = ctrl.Antecedent(np.arange(0, 101, 1), 'margin')
competitor_discount = ctrl.Antecedent(np.arange(0, 101, 1), 'competitor_discount')
discount = ctrl.Consequent(np.arange(0, 71, 1), 'discount')

# huấn luyện
rating['thap'] = fuzz.trimf(rating.universe, [0, 0, 4.0])
rating['trung_binh'] = fuzz.trimf(rating.universe, [4.0, 4.25, 4.5])
rating['cao'] = fuzz.trimf(rating.universe, [4.5, 4.8, 5.0])

sales['it'] = fuzz.trimf(sales.universe, [0, 0, 400])
sales['cao'] = fuzz.trimf(sales.universe, [500, 1000, 1000])

margin['thap'] = fuzz.trimf(margin.universe, [0, 0, 30])
margin['cao'] = fuzz.trimf(margin.universe, [60, 100, 100])

competitor_discount['thap'] = fuzz.trimf(competitor_discount.universe, [0, 0, 50])
competitor_discount['cao'] = fuzz.trimf(competitor_discount.universe, [40, 100, 100])

discount['rat_thap'] = fuzz.trimf(discount.universe, [0, 2.5, 5])
discount['thap'] = fuzz.trimf(discount.universe, [5, 7.5, 10])
discount['trung_binh'] = fuzz.trimf(discount.universe, [10, 15, 20])
discount['cao'] = fuzz.trimf(discount.universe, [20, 30, 40])
discount['rat_cao'] = fuzz.trimf(discount.universe, [40, 55, 70])
#logic mờ
rule1 = ctrl.Rule(competitor_discount['cao'] & margin['cao'], discount['rat_cao'])
rule2 = ctrl.Rule(competitor_discount['thap'] & rating['cao'], discount['rat_thap'])
rule3 = ctrl.Rule(margin['thap'] & competitor_discount['cao'], discount['thap'])
rule4 = ctrl.Rule(rating['thap'] & sales['it'], discount['cao'])
rule5 = ctrl.Rule(margin['cao'] & sales['it'], discount['trung_binh'])
rule6 = ctrl.Rule(rating['trung_binh'] & competitor_discount['cao'], discount['cao'])
rule7 = ctrl.Rule(competitor_discount['thap'] & sales['cao'] & rating['cao'], discount['rat_thap'])


shopee_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6, rule7])
shopee_sim = ctrl.ControlSystemSimulation(shopee_ctrl)

# nhập dữ liệu
shopee_sim.input['rating'] = 4.2
shopee_sim.input['sales'] = 300
shopee_sim.input['margin'] = 70
shopee_sim.input['competitor_discount'] = 60
shopee_sim.compute()

print(f"Tỷ lệ chiết khấu đề xuất sau khi xét đối thủ: {shopee_sim.output['discount']:.2f}%")
```

```python
#2.13
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

#vao
demand = ctrl.Antecedent(np.arange(0, 101, 1), 'demand')
comp_pressure = ctrl.Antecedent(np.arange(0, 101, 1), 'comp_pressure')
profit_margin = ctrl.Antecedent(np.arange(0, 101, 1), 'profit_margin')
reputation = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'reputation')
seasonal_demand = ctrl.Antecedent(np.arange(0, 101, 1), 'seasonal_demand')

#ra
discount = ctrl.Consequent(np.arange(0, 71, 1), 'discount')

for x in [demand, comp_pressure, profit_margin, seasonal_demand]:
    x['thap'] = fuzz.trimf(x.universe, [0, 0, 50])
    x['trung_binh'] = fuzz.trimf(x.universe, [30, 50, 70])
    x['cao'] = fuzz.trimf(x.universe, [50, 100, 100]) #( đã sửa bởi gmn)


reputation['thap'] = fuzz.trimf(reputation.universe, [0, 0, 4])
reputation['trung_binh'] = fuzz.trimf(reputation.universe, [4, 4.25, 4.5])
reputation['cao'] = fuzz.trimf(reputation.universe, [4.5, 5, 5])

discount['rat_thap'] = fuzz.trimf(discount.universe, [0, 2.5, 5])
discount['thap'] = fuzz.trimf(discount.universe, [5, 7.5, 10])
discount['trung_binh'] = fuzz.trimf(discount.universe, [10, 15, 20])
discount['cao'] = fuzz.trimf(discount.universe, [20, 30, 40])
discount['rat_cao'] = fuzz.trimf(discount.universe, [40, 55, 70])


rule1 = ctrl.Rule(demand['cao'] & comp_pressure['thap'] & profit_margin['cao'],
    discount['rat_thap'])

rule2 = ctrl.Rule( demand['thap'] & comp_pressure['cao'] & profit_margin['cao'],
    discount['cao'])

rule3 = ctrl.Rule( reputation['cao'] & profit_margin['trung_binh'] & seasonal_demand['cao'],
    discount['trung_binh'])

rule4 = ctrl.Rule( comp_pressure['cao'] & seasonal_demand['cao'] & profit_margin['cao'],
    discount['rat_cao'])

rule5 = ctrl.Rule( reputation['thap'] & demand['trung_binh'] & profit_margin['thap'],
    discount['trung_binh'])

rule6 = ctrl.Rule(demand['cao'] & seasonal_demand['thap'] & comp_pressure['thap'],
    discount['rat_thap'])

rule7 = ctrl.Rule(profit_margin['cao'] & comp_pressure['trung_binh'] & seasonal_demand['trung_binh'],
    discount['trung_binh'])

rule8 = ctrl.Rule(demand['cao'] & comp_pressure['trung_binh'] & reputation['trung_binh'] & profit_margin['cao'] & seasonal_demand['cao'],
    discount['trung_binh']) # luật từ đề bài cho


system_ctrl = ctrl.ControlSystem([
    rule1, rule2, rule3, rule4,
    rule5, rule6, rule7, rule8])

system = ctrl.ControlSystemSimulation(system_ctrl)
# tình huống :
system.input['demand'] = 80
system.input['comp_pressure'] = 50
system.input['reputation'] = 4.2
system.input['profit_margin'] = 85
system.input['seasonal_demand'] = 90

system.compute()

print(" KẾT QUẢ ĐỊNH GIÁ")
print("San pham: Dong ho xa xi thu cong")
print(" chiet khau de xuat:", round(system.output['discount'], 2), "%")
```
